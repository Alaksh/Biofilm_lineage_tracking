```python
import subprocess
import os

split_files = os.listdir("Fastq/")
def usearch_merge(file_name,sample_name):
    p = subprocess.call(["usearch10.0.240_win32.exe","-fastq_mergepairs",file_name,\
                         "-fastqout",'Merged/'+sample_name+'.fq',"-sample","Sample"+sample_name])
    return p

i = 1
for files in split_files:
    if ('R1' in files) and ('.gz' not in files):
        input_file = 'Fastq/'+files
        output_file = files.split('_')[0]
        print(output_file)
        p = usearch_merge(input_file,output_file)
        if p == 0:
            print("Usearch completed successfuly for -" + input_file)
        else:
            print("Usearch not done for -" + input_file)
        i = i +1
```

```python
"""
The following code is to execute the usearch function. 
"""
import subprocess
import os

split_files = os.listdir("Merged/")
#split_files.remove('desktop.ini')
def usearch_function(file_name,output_filename):
    userfields = "query+target+id+alnlen+mism+opens+qlo+qhi+tlo+thi+evalue+bits+qrow+trow+qrowdots+tstrand+qstrand"
    p = subprocess.call(["usearch10.0.240_win32.exe","-usearch_global",file_name,"-db","Template.txt","-id","0.45",\
                        "-strand","both","-userout",output_filename,"-userfields",\
                        userfields,"-maxaccepts","40","-maxhits","1","-mincols","95","-maxrejects","35"])
    return p

print(len(split_files))
#split_files.remove('desktop.ini')
for files in split_files:
    input_file = 'Merged/'+files
    fine_name = files
    output_file = 'Usearch/'+fine_name+'.txt'
    print(output_file)
    p = usearch_function(input_file,output_file)
    if p == 0:
        print("Usearch completed successfuly for -" + input_file)
    else:
       print("Usearch not done for -" + input_file)
```
